```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("../data/raw/ai_jobs_market_2025_2026.csv")
df
```

```python
df.info()
```

```python
clean_df = df.copy()
```

```python
print("Rows:",clean_df.shape[0])
print("Columns:",clean_df.shape[1])
```

```python
clean_df.info()
```

```python
clean_df.columns
```

```python
clean_df.columns = (
    clean_df.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
)
clean_df.columns
```

```python
text_columns = clean_df.select_dtypes(include="object").columns
text_columns
```

```python
for col in text_columns:
    clean_df[col] = clean_df[col].str.strip()
```

```python
# Missing Values ka Percentage
missing_percentage = (
    clean_df.isnull().sum() / len(clean_df) * 100
)
missing_percentage[missing_percentage > 0]
```

```python
# Missing values handle 
clean_df["education_required"] = (
    clean_df["education_required"]
    .fillna("Unknown")
)
clean_df
```

## Duplicate Records


```python
clean_df.duplicated().sum()
```

```python
clean_df["job_id"].duplicated().sum()
```

```python
# Data types fix 
numeric_columns = [
    "years_of_experience",
    "annual_salary_usd",
    "salary_min_usd",
    "salary_max_usd",
    "ai_salary_premium_pct",
    "demand_score",
    "demand_growth_yoy_pct",
    "benefits_score_10",
    "posting_year"
]
```

```python
for col in numeric_columns:
    clean_df[col] = pd.to_numeric(
        clean_df[col],
        errors="coerce"
    )

clean_df[numeric_columns].dtypes
```

```python
# Boolean Columns Check
boolean_columns = [
    "is_senior",
    "is_remote_friendly",
    "is_llm_role"
]

for col in boolean_columns:
    print()
    print(clean_df[col].value_counts(dropna=False))
```

```python
# Invalid Salary Values
clean_df.loc[
    
    clean_df["annual_salary_usd"] < 0,
    "annual_salary_usd"
] = np.nan
clean_df
```

```python
# Experience validate 
clean_df.loc[
    clean_df["years_of_experience"] < 0,
    "years_of_experience"
] = np.nan
clean_df
```

```python
# Benefits Score Validate 
clean_df.loc[
    (clean_df["benefits_score_10"] < 0) |
    (clean_df["benefits_score_10"] < 0),
    "benefits_score_10"
] = np.nan

clean_df["benefits_score_10"]
```

## Categorical Values Standardize karna 


```python
clean_df["country"].value_counts()
```

```python
# Text cleanup
clean_df["country"] = (
    clean_df["country"]
    .str.strip()
    .str.title()
)
clean_df["country"]
```

```python
# Remote Work
clean_df["remote_work"].value_counts()
```

```python
clean_df["remote_work"] = (
    clean_df["remote_work"]
    .str.strip()
    .str.title()
)
clean_df["remote_work"]
```

```python
## Experience Level
clean_df["experience_level"].value_counts()
```

```python
clean_df["experience_level"] = (
    clean_df["experience_level"]
    .str.strip()
)
clean_df["experience_level"]
```

```python
# Job title
clean_df["job_category"].value_counts()
```

```python
clean_df["job_category"] = (
    clean_df["job_category"]
    .str.strip()
)
clean_df["job_category"]
```

```python
# Industry
clean_df["industry"].value_counts()
```

```python
clean_df["industry"] = (
    clean_df["industry"] 
    .str.strip()
)
clean_df["industry"] 
```

```python
# Outliers
Q1 = clean_df["annual_salary_usd"].quantile(0.25)
Q3 = clean_df["annual_salary_usd"].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR
```

```python
salary_outliers = clean_df[
    (clean_df["annual_salary_usd"] < lower_bound) |
    (clean_df["annual_salary_usd"] > upper_bound)
]

salary_outliers
```
